# Imitation Learning - ACT

Imitation learning is a machine learning approach where a model is trained to mimic expert behavior by observing and replicating demonstrations, enabling it to perform tasks similarly to the expert. ACT is an action chunking policy with Transformers, an architecture designed for sequence modeling, and train it as a conditional VAE (CVAE) to capture the variability in human data. It significantly outperforms previous imitation learning algorithms on a range of simulated and real-world fine manipulation tasks. See the original paper for details: [Action Chunking Transformer](https://arxiv.org/pdf/2304.13705).

We have built an imitation learning pipeline for ACT, which can be used to train and evaluate the ACT model on different tasks. In this tutorial, we will introduce how to setup ACT pipeline from installing all the required packages to convert the model to OpenVINO and running the model.

**Note:** This notebook currently supports **Ubuntu 22.04** only.

## Convert PyTorch Model Checkpoint to OpenVINO IR

The first conversion cell executes `ov_convert.py` and creates the default OpenVINO IR with FP16-compressed weights. The following cell additionally exports an FP32-weight IR and an INT8 weight-compressed IR. INT8 compression is weight-only because representative calibration data is not included with the checkpoint package; model inputs and activations remain floating point.

The first cell runs the following command:

`python3 ov_convert.py --ckpt_path <ckpt_path> --height 480 --weight 640 --camera_num 4 --chunk_size 100`

```python
import os
import sys
import subprocess
import pathlib

CKPT_PATH = pathlib.Path("sim_insertion_scripted/four_camera/policy_best.ckpt").resolve()
ACT_DIR = pathlib.Path("edge-ai-suites/robotics-ai-suite/pipelines/act-sample/act").resolve()
SCRIPT = ACT_DIR / "ov_convert.py"


if not SCRIPT.exists():
    raise FileNotFoundError(f"ov_convert.py not found: {SCRIPT}")
if not CKPT_PATH.exists():
    raise FileNotFoundError(f"Checkpoint not found: {CKPT_PATH}")

cmd = [
    sys.executable,
    str(SCRIPT),
    "--ckpt_path",
    str(CKPT_PATH),
    "--height",
    "480",
    "--weight",
    "640",
    "--camera_num",
    "4",
    "--chunk_size",
    "100",
]
print("[STEP] Running from:", ACT_DIR)
print("[STEP] Command:", " ".join(cmd))

# Create environment with ACT_DIR in PYTHONPATH
env = os.environ.copy()
env["PYTHONPATH"] = str(ACT_DIR) + os.pathsep + env.get("PYTHONPATH", "")
print("[INFO] PYTHONPATH:", env["PYTHONPATH"])

res = subprocess.run(cmd, capture_output=True, text=True, cwd=str(ACT_DIR), env=env)
print(res.stdout)
if res.returncode != 0:
    print(res.stderr)
    raise subprocess.CalledProcessError(res.returncode, cmd)
```

```python
# Export additional FP32 and fully quantized INT8 OpenVINO models.
# INT8 PTQ calibrates weights and activations with representative simulator observations.
import pickle
import subprocess
import sys

import nncf
import numpy as np
import openvino as ov

FP32_PATH = CKPT_PATH.with_name(f"{CKPT_PATH.stem}_fp32.xml")
INT8_PATH = CKPT_PATH.with_name(f"{CKPT_PATH.stem}_int8.xml")
STATS_PATH = CKPT_PATH.with_name("dataset_stats.pkl")
CALIBRATION_SAMPLES = 128
VALIDATION_SAMPLES = 8
CALIBRATION_EPISODES = 3
EPISODE_LENGTH = 400
ACTION_HORIZON = 100
CAMERA_NAMES = ["top", "angle", "left_wrist", "right_wrist"]

# ov_convert.py calls ov.save_model() with its default FP16 compression. Run it
# through a small wrapper that disables that compression to preserve FP32 weights.
if not FP32_PATH.exists():
    fp32_wrapper = """
import runpy
import sys
import openvino as ov

original_save_model = ov.save_model
ov.save_model = lambda model, output_model: original_save_model(model, output_model, compress_to_fp16=False)
script = sys.argv[1]
sys.argv = sys.argv[1:]
runpy.run_path(script, run_name="__main__")
"""
    fp32_cmd = [
        sys.executable,
        "-c",
        fp32_wrapper,
        str(SCRIPT),
        "--ckpt_path",
        str(CKPT_PATH),
        "--output_name",
        str(FP32_PATH),
        "--height",
        "480",
        "--weight",
        "640",
        "--camera_num",
        "4",
        "--chunk_size",
        str(ACTION_HORIZON),
    ]
    print("[STEP] Exporting FP32 model...")
    subprocess.run(fp32_cmd, check=True, cwd=str(ACT_DIR), env=env)
else:
    print(f"[INFO] Reusing FP32 model: {FP32_PATH}")

if not STATS_PATH.exists():
    raise FileNotFoundError(f"Dataset statistics not found: {STATS_PATH}")

# Import the simulator after adding ACT_DIR to the module search path.
if str(ACT_DIR) not in sys.path:
    sys.path.insert(0, str(ACT_DIR))
from sim_env import BOX_POSE, make_sim_env
from utils import sample_insertion_pose

with STATS_PATH.open("rb") as file:
    stats = pickle.load(file)

def prepare_sample(sample):
    """Convert compact simulator data to the two FP32 model inputs."""
    return {
        "qpos": sample["qpos"][np.newaxis, :].astype(np.float32),
        "tensor.1": sample["images"][np.newaxis, :].astype(np.float32) / 255.0,
    }

def collect_calibration_data(policy, sample_count, episode_count=CALIBRATION_EPISODES):
    """Collect compact samples across complete FP32 policy-guided rollouts."""
    samples = []
    samples_per_episode = int(np.ceil(sample_count / episode_count))

    for episode_idx in range(episode_count):
        np.random.seed(episode_idx)
        BOX_POSE[0] = np.concatenate(sample_insertion_pose())
        sim = make_sim_env("sim_insertion_scripted")
        timestep = sim.reset()
        sample_steps = set(np.linspace(0, EPISODE_LENGTH - 1, samples_per_episode, dtype=int))
        action_chunk = None

        for step in range(EPISODE_LENGTH):
            observation = timestep.observation
            qpos = (np.asarray(observation["qpos"], dtype=np.float32) - stats["qpos_mean"]) / stats["qpos_std"]
            images = np.stack([observation["images"][name] for name in CAMERA_NAMES])
            images = images.transpose(0, 3, 1, 2)  # Retain uint8 to limit calibration memory.
            model_input = prepare_sample({"qpos": qpos, "images": images})

            if step in sample_steps and len(samples) < sample_count:
                samples.append({"qpos": qpos.astype(np.float32), "images": images.copy()})

            if step % ACTION_HORIZON == 0:
                action_chunk = policy(model_input)[0]
            normalized_action = action_chunk[0, step % ACTION_HORIZON]
            action = normalized_action * stats["action_std"] + stats["action_mean"]
            timestep = sim.step(action)

        del sim
        if len(samples) >= sample_count:
            break

    if len(samples) != sample_count:
        raise RuntimeError(f"Collected {len(samples)} of {sample_count} requested samples")
    return samples

# Always regenerate this path: an older file may contain weight-only compression.
fp32_model = ov.Core().read_model(FP32_PATH)
fp32_policy = ov.compile_model(fp32_model, "CPU")
total_samples = CALIBRATION_SAMPLES + VALIDATION_SAMPLES
print(f"[STEP] Collecting {total_samples} representative simulator samples...")
representative_data = collect_calibration_data(fp32_policy, total_samples)
calibration_data = representative_data[:CALIBRATION_SAMPLES]
validation_data = representative_data[CALIBRATION_SAMPLES:]
calibration_dataset = nncf.Dataset(calibration_data, prepare_sample)

print("[STEP] Running NNCF static INT8 quantization (weights + activations)...")
int8_model = nncf.quantize(
    fp32_model,
    calibration_dataset,
    preset=nncf.QuantizationPreset.PERFORMANCE,
    target_device=nncf.TargetDevice.ANY,
    model_type=nncf.ModelType.TRANSFORMER,
    subset_size=len(calibration_data),
)
ov.save_model(int8_model, INT8_PATH, compress_to_fp16=False)

# FakeQuantize nodes distinguish static PTQ from weight-only compression.
fake_quantize_count = sum(op.get_type_name() == "FakeQuantize" for op in int8_model.get_ops())
if fake_quantize_count == 0:
    raise RuntimeError("NNCF produced no FakeQuantize operations; the model is not statically quantized")

# Check quantization error on simulator samples excluded from calibration.
int8_policy = ov.compile_model(int8_model, "CPU")
absolute_errors = []
for sample in validation_data:
    model_input = prepare_sample(sample)
    fp32_output = fp32_policy(model_input)[0]
    int8_output = int8_policy(model_input)[0]
    absolute_errors.append(np.abs(fp32_output - int8_output))
absolute_errors = np.concatenate([error.ravel() for error in absolute_errors])

for precision, model_path in {"FP32": FP32_PATH, "INT8 PTQ": INT8_PATH}.items():
    model_size = model_path.stat().st_size + model_path.with_suffix(".bin").stat().st_size
    print(f"[DONE] {precision}: {model_path} ({model_size / 1024**2:.2f} MiB)")
print(f"[DONE] INT8 FakeQuantize operations: {fake_quantize_count}")
print(f"[CHECK] Held-out output MAE: {absolute_errors.mean():.6f}; max error: {absolute_errors.max():.6f}")
```

## Evaluate the Policy

Run the next cell to evaluate the FP16, FP32, and statically quantized INT8 models. Each model runs the same 10 deterministic rollout episodes so their success rates and average returns can be compared fairly. This rollout evaluation is the primary task-accuracy check for the calibrated INT8 model.

The cell executes `imitate_episodes.py` with the following command for each precision:

`python3 imitate_episodes.py --task_name sim_insertion_scripted --ckpt_dir <precision_ckpt_dir> --policy_class ACT --kl_weight 10 --chunk_size 100 --hidden_dim 512 --batch_size 8 --dim_feedforward 3200 --num_epochs 2000 --lr 1e-5 --seed 0 --device CPU/GPU/NPU --eval`

**Note:** Evaluating all three models may take 30 to 60 minutes depending on your hardware.

Below is a demonstration of the peg insertion task:

![Peg Insertion Demo](https://github.com/open-edge-platform/edge-ai-suites/raw/main/robotics-ai-suite/pipelines/act-sample/README.assets/act-sim-insertion-demo.gif)

### Understanding the Output

Each precision runs 10 rollout episodes and reports the following metrics:

**Per-Episode Metrics:**
- **`episode_return`**: Total cumulative reward for the entire episode (sum of all step rewards)
- **`episode_highest_reward`**: Maximum reward achieved in a single step during the episode
- **`env_max_reward`**: Maximum possible reward per step (4 for this task)
- **`Success`**: Whether the episode met the success criteria
- **`Avg fps`**: Inference speed (frames per second)

**Overall Results:**
- **Success rate**: Percentage of episodes that completed successfully
- **Average return**: Mean cumulative reward across all episodes
- **Reward breakdown**: Hierarchical achievement levels (0-4), where:
  - Reward >= 0: Basic interaction
  - Reward >= 1-3: Partial task completion
  - Reward >= 4: Full task completion (successful peg insertion)

A perfect run achieves 100% success rate with all episodes reaching reward level 4.

```python
import os
import re
import shutil
import sys
import subprocess
import pathlib

CKPT_DIR = pathlib.Path("sim_insertion_scripted/four_camera").resolve()
ACT_DIR = pathlib.Path("edge-ai-suites/robotics-ai-suite/pipelines/act-sample/act").resolve()
SCRIPT = ACT_DIR / "imitate_episodes.py"

if not SCRIPT.exists():
    raise FileNotFoundError(f"imitate_episodes.py not found: {SCRIPT}")
if not CKPT_DIR.exists():
    raise FileNotFoundError(f"Checkpoint directory not found: {CKPT_DIR}")

MODEL_PATHS = {
    "FP16": CKPT_DIR / "policy_best.xml",
    "FP32": CKPT_DIR / "policy_best_fp32.xml",
    "INT8": CKPT_DIR / "policy_best_int8.xml",
}
for precision, model_path in MODEL_PATHS.items():
    if not model_path.exists() or not model_path.with_suffix(".bin").exists():
        raise FileNotFoundError(f"{precision} OpenVINO model not found: {model_path}")

base_cmd = [
    sys.executable,
    str(SCRIPT),
    "--task_name",
    "sim_insertion_scripted",
    "--policy_class",
    "ACT",
    "--kl_weight",
    "10",
    "--chunk_size",
    "100",
    "--hidden_dim",
    "512",
    "--batch_size",
    "8",
    "--dim_feedforward",
    "3200",
    "--num_epochs",
    "2000",
    "--lr",
    "1e-5",
    "--seed",
    "0",
    "--device",
    TARGET_DEVICE.value,
    "--eval",
]

# Create environment with ACT_DIR in PYTHONPATH.
env = os.environ.copy()
env["PYTHONPATH"] = str(ACT_DIR) + os.pathsep + env.get("PYTHONPATH", "")

# imitate_episodes.py always loads policy_best.xml. Create one isolated checkpoint
# directory per precision so the same evaluation path is used without renaming
# or overwriting the exported models.
if "DISPLAY" not in env:
    print("[INFO] No display detected (headless). Skipping evaluation cell.")
else:
    evaluation_results = {}
    for precision, model_path in MODEL_PATHS.items():
        precision_dir = CKPT_DIR / f"evaluation_{precision.lower()}"
        precision_dir.mkdir(exist_ok=True)
        shutil.copy2(model_path, precision_dir / "policy_best.xml")
        shutil.copy2(model_path.with_suffix(".bin"), precision_dir / "policy_best.bin")
        shutil.copy2(CKPT_DIR / "dataset_stats.pkl", precision_dir / "dataset_stats.pkl")

        cmd = [*base_cmd, "--ckpt_dir", str(precision_dir)]
        print(f"\n[STEP] Evaluating {precision} model...")
        res = subprocess.run(cmd, capture_output=True, text=True, cwd=str(ACT_DIR), env=env)
        print(res.stdout)
        if res.returncode != 0:
            print(res.stderr)
            raise subprocess.CalledProcessError(res.returncode, cmd)

        success_match = re.search(r"Success rate:\s*([0-9.]+)", res.stdout)
        return_match = re.search(r"Average return:\s*([-+0-9.eE]+)", res.stdout)
        if success_match is None or return_match is None:
            raise RuntimeError(f"Could not parse {precision} evaluation metrics")
        evaluation_results[precision] = {
            "success_rate": float(success_match.group(1)),
            "average_return": float(return_match.group(1)),
        }

    print("\nPrecision comparison (10 rollouts per model):")
    print(f"{'Precision':<10} {'Success rate':>14} {'Average return':>16}")
    for precision, metrics in evaluation_results.items():
        print(f"{precision:<10} {metrics['success_rate']:>13.1%} {metrics['average_return']:>16.3f}")
```
